`Remove Charged Species`

```python
def remove_charged_species_smiles(input_path, output_path):
    kept = 0
    removed = 0

    with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
        for line in file:
            parts = line.strip().split(None, 1)
            if len(parts) != 2:
                continue
            _, smiles = parts

            if '+' in smiles or '-' in smiles:
                removed += 1
                continue  

            output.write(line)
            kept += 1

    print(f"Removed SMILES (charged species): {removed}")
    print(f"Total SMILES: {kept}")

input_file = "./Sources/CID-SMILES"       
output_file = "./Sources/PubChemDatasets/CID-SMILES-NoChargedSpecies.txt"  

remove_charged_species_smiles(input_file, output_file)
```

`Remove Metallic Compounds`

```python
def remove_metallic_smiles(input_path, output_path):
    metals = [
        "Li", "Na", "K", "Rb", "Cs", "Fr",
        "Be", "Mg", "Ca", "Sr", "Ba", "Ra",
        "Fe", "Cu", "Zn", "Ni", "Au", "Ag", "Hg",
        "La", "Lu", "Ac", "Lr",
        "Al", "Ga", "In", "Sn", "Tl", "Pb", "Bi", "Po"
        "Sc", "Y", "Ti", "Zr", "Hf", "Rf",
        "V", "Nb", "Ta", "Db",
        "Cr", "Mo", "W", "Sg",
        "Mn", "Tc", "Re", "Bh",
        "Ru", "Os", "Hs",
        "Co", "Rh", "Ir", "Mt",
        "Pd", "Pt", "Ds",
        "Rg", "Cd", "Cn",
        "Nh", "Fl", "Mc", "Lv"
    ]
    
    kept = 0
    removed = 0

    with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
        for line in file:
            parts = line.strip().split(None, 1)
            if len(parts) != 2:
                continue
            _, smiles = parts

            if any(metal in smiles for metal in metals):
                removed += 1
                continue

            output.write(line)
            kept += 1

    print(f"Removed SMILES (with metals): {removed}")
    print(f"Total SMILES: {kept}")

input_file = "./Sources/PubChemDatasets/CID-SMILES-NoChargedSpecies.txt"       
output_file = "./Sources/PubChemDatasets/CID-SMILES-NoMetallicCompounds.txt"             

remove_metallic_smiles(input_file, output_file)
```

`Remove Halogenated Compounds`

```python
def remove_halogens_smiles(input_path, output_path):
    halogens = ["F", "Cl", "Br", "I", "At", "Ts"]
    kept = 0
    removed = 0

    with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
        for line in file:
            parts = line.strip().split(None, 1)
            if len(parts) != 2:
                continue
            _, smiles = parts

            if any(halogen in smiles for halogen in halogens):
                removed += 1
                continue

            output.write(line)
            kept += 1

    print(f"Removed SMILES (halogens): {removed}")
    print(f"Total SMILES: {kept}")

input_file = "./Sources/PubChemDatasets/CID-SMILES-NoMetallicCompounds.txt"      
output_file = "./Sources/PubChemDatasets/CID-SMILES-NoHalogenatedCompounds.txt"  

remove_halogens_smiles(input_file, output_file)
```

`Remove Mixtures SMILES`

```python
def remove_mixtures_smiles(input_path, output_path):
    kept = 0
    removed = 0

    with open(input_path, 'r', encoding='utf-8') as file, open(output_path, 'w', encoding='utf-8') as output:
        for line in file:
            parts = line.strip().split(None, 1)
            if len(parts) != 2:
                continue
            _, smiles = parts

            if '.' in smiles:
                removed += 1
                continue  

            output.write(line)
            kept += 1

    print(f"Removed SMILES (with mixtures): {removed}")
    print(f"Total SMILES: {kept}")

input_file = "./Sources/PubChemDatasets/CID-SMILES-NoHalogenatedCompounds.txt"  
output_file = "./Sources/PubChemDatasets/CID-SMILES-NoMixtures.txt"  

remove_mixtures_smiles(input_file, output_file)
```

### Data Sources

- **`CRC_ORG`**  
  Compilation of evaluated data on organic compounds from the CRC Handbook of Chemistry and Physics [1].

- **`CRC_INORG`**  
  Compilation of evaluated data on inorganic compounds from the CRC Handbook of Chemistry and Physics [1].

- **`YAWS`**  
  Compilation of experimental and predicted data aggregated from multiple sources, without individual sources attribution [2].

- **`WIKIDATA`**  
  Property values retrieved from the collaborative Wikidata knowledge base via its public API [3].

- **`COMMON_CHEMISTRY`**  
  Curated data from the CAS Common Chemistry project [4].

- **`JOBACK`**  
  Properties estimated for organic compounds via the Joback-Reid group-contribution method [5].

- **`WEBBOOK`**  
  Compiled experimental thermochemical and thermophysical data from the NIST Chemistry WebBook [6].

- **`HEOS`**  
  Thermophysical properties computed from fundamental Helmholtz-energy equations of state, as distributed in the NIST REFPROP Database [7].

- Retrieve all melting points from sources

```python
mp_columns = ['Tm']
records = []

for source, df in Tm_sources.items():
    mp_col = next((col for col in mp_columns if col in df.columns), None)
   
    if 'CASRN' in df.columns:
        result_df = df[['CASRN', mp_col]].copy()
    else:
        result_df = df[[mp_col]].copy()
        result_df.reset_index(inplace=True)
        result_df.rename(columns={'index': 'CASRN'}, inplace=True)

    result_df.columns = ['CASRN', 'Tm']
    result_df['Source'] = source
    records.append(result_df)

all_melting_points = pd.concat(records, ignore_index=True)
all_melting_points = all_melting_points.dropna(subset=['Tm'])
all_melting_points = all_melting_points.drop_duplicates(subset='CASRN')
print(f"Total unique melting point records: {len(all_melting_points)}")
```

- Retrieve all boiling points from sources

```python
bp_columns = ['Tb', 'T', 'T_b']
records = []

for source, df in Tb_sources.items():
    bp_col = next((col for col in bp_columns if col in df.columns), None)
    
    if 'CASRN' in df.columns:
        result_df = df[['CASRN', bp_col]].copy()
    else:
        result_df = df[[bp_col]].copy()
        result_df.reset_index(inplace=True)
        result_df.rename(columns={'index': 'CASRN'}, inplace=True)

    result_df.columns = ['CASRN', 'Tb'] 
    result_df['Source'] = source
    records.append(result_df)

all_boiling_points = pd.concat(records, ignore_index=True)
all_boiling_points = all_boiling_points.dropna(subset=['Tb'])
all_boiling_points = all_boiling_points.drop_duplicates(subset='CASRN')
print(f"Total unique boiling point records: {len(all_boiling_points)}")
```

```python
def get_cids_from_cas(cas_list):
    cid_results = {}
    total = len(cas_list)
    for i, cas in enumerate(cas_list, 1):
        try:
            cids = pcp.get_cids(cas, 'name')
            if cids:
                cid_results[cas] = cids[0]
            else:
                cid_results[cas] = None
        except Exception:
            cid_results[cas] = None

        cid = cid_results[cas]
        cid_str = f"CID: {cid}" if cid else "CID not found"
        print(f"[{i}/{total}] CAS: {cas} - {cid_str}")

    return cid_results
```

`Keep compounds with H, C, N, O, S`

```python
allowed_elements = {'H', 'C', 'N', 'O', 'S'}

def contains_only_HCNOS(smiles):
    elements = re.findall(r'\[.*?\]|Br|Cl|[A-Z][a-z]?', smiles)

    cleaned_elements = []
    for el in elements:
        if el.startswith('['):
            el = el.strip('[]')
            sub_elements = re.findall(r'[A-Z][a-z]?', el)
            cleaned_elements.extend(sub_elements)
        else:
            cleaned_elements.append(el)

    return all(el in allowed_elements for el in cleaned_elements)

data = data[data['SMILES'].apply(contains_only_HCNOS)].copy()
data.shape
```

`1) Filter by Temperature` - Keep records from CRC_ORG or CRC_INORG

```python
EXPERIMENTAL_PRIORITY = {'CRC_ORG', 'CRC_INORG'}

def filter_prefer_crc(df):
    removed = 0
    kept_rows = []

    for _, group in df.groupby('CAS'):
        if len(group) == 1:
            kept_rows.append(group)
            continue

        crc_rows = group[
            group['Tb Source'].isin(EXPERIMENTAL_PRIORITY) |
            group['Tm Source'].isin(EXPERIMENTAL_PRIORITY)
        ]
        if not crc_rows.empty:
            removed += len(group) - 1
            kept_rows.append(crc_rows.iloc[[0]])
        else:
            kept_rows.append(group)

    return pd.concat(kept_rows), removed

data, removed_crc = filter_prefer_crc(data)
print(f"Current Number records: {len(data)} vs Original Number records: {original_number_records}")
print(f"Removed records: {removed_crc}")
```

` 2) Filter by Temperature` - If only computed data exists (JOBACK) keep first instance

```python
def filter_all_joback_only(df):
    removed = 0
    kept_rows = []

    for _, group in df.groupby('CAS'):
        if (group['Tb Source'] == 'JOBACK').all() and (group['Tm Source'] == 'JOBACK').all():
            removed += len(group) - 1
            kept_rows.append(group.iloc[[0]])
        else:
            kept_rows.append(group)
    
    return pd.concat(kept_rows), removed

data, removed_joback = filter_all_joback_only(data)
print(f"Current Number records: {len(data)} vs Original Number records: {original_number_records}")
print(f"Removed records: {removed_joback}")
```

`3) Filter by Temperature` - Keep only first instance

```python
def filter_keep_first_only(df):
    removed = 0
    kept_rows = []

    for _, group in df.groupby('CAS'):
        if len(group) > 1:
            removed += len(group) - 1
        kept_rows.append(group.iloc[[0]])

    return pd.concat(kept_rows), removed

data, removed_final = filter_keep_first_only(data)
print(f"Current Number records: {len(data)} vs Original Number records: {original_number_records}")
print(f"Removed records: {removed_final}")
```

```python
EXCLUDED_FRAGMENTS = [
    Fragments.fr_C_S,           
    Fragments.fr_Imine,         
    Fragments.fr_NH1,           
    Fragments.fr_NH2,           
    Fragments.fr_SH,            
    Fragments.fr_aniline,       
    Fragments.fr_azide,         
    Fragments.fr_azo,           
    Fragments.fr_barbitur,      
    Fragments.fr_benzodiazepine,
    Fragments.fr_diazo,
    Fragments.fr_epoxide,
    Fragments.fr_furan,
    Fragments.fr_guanido,
    Fragments.fr_hdrzine,       
    Fragments.fr_hdrzone,       
    Fragments.fr_imidazole,
    Fragments.fr_imide,
    Fragments.fr_isocyan,       
    Fragments.fr_isothiocyan,
    Fragments.fr_nitro,
    Fragments.fr_nitro_arom,
    Fragments.fr_nitro_arom_nonortho,
    Fragments.fr_nitroso,
    Fragments.fr_oxazole,
    Fragments.fr_oxime,
    Fragments.fr_phos_acid,
    Fragments.fr_phos_ester,
    Fragments.fr_piperzine,
    Fragments.fr_piperdine,
    Fragments.fr_prisulfonamd,
    Fragments.fr_sulfide,
    Fragments.fr_sulfonamd,
    Fragments.fr_term_acetylene,
    Fragments.fr_tetrazole,
    Fragments.fr_thiazole,
    Fragments.fr_thiocyan,
    Fragments.fr_thiophene,
    Fragments.fr_urea,
    Fragments.fr_N_O,
    Fragments.fr_Nhpyrrole,
    Fragments.fr_alkyl_carbamate,
    Fragments.fr_lactam,
    Fragments.fr_morpholine,
    Fragments.fr_quatN,
    Fragments.fr_aldehyde,
    Fragments.fr_amidine,
    Fragments.fr_Ar_NH,
    Fragments.fr_dihydropyridine     
]

def has_excluded_fragments(smiles):
    mol = Chem.MolFromSmiles(smiles)
    return any(fragment(mol) > 0 for fragment in EXCLUDED_FRAGMENTS)

initial = len(data)
df_filtered = data[~data['SMILES'].apply(has_excluded_fragments)].copy()
print(f"Before: {initial} | After: {len(df_filtered)} | Removed: {initial - len(df_filtered)}")
```

`Filter by Adjacent Functional Groups` - SMARTS

```python
excluded_patterns = {"adjacent O-O" : Chem.MolFromSmarts("[O][O]"), "adjacent S-S" : Chem.MolFromSmarts("[S][S]")}

def has_excluded_patterns(smiles):
    mol = Chem.MolFromSmiles(smiles)
    return any(mol.HasSubstructMatch(pattern)for pattern in excluded_patterns.values())

initial = len(df_filtered)
df_results = df_filtered[~df_filtered['SMILES'].apply(has_excluded_patterns)].copy()
print(f"Before: {initial} | After: {len(df_results)} | Removed: {initial - len(df_results)}")               
```
